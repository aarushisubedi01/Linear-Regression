### **Linear Regression**


```python
# One of the simplest yet powerful Machine Learning Algorithm
# It is used for predicting a continuous target variable based one or more features variables.

# Linear Regression finds the best-fitting straight line(regression line)
        # y = mx + c
# y = target variable (dependent variable)
# x= features (independent variables)
# m = slope
# c = intercept ( Simple Linear Regression)

# Multiple linear Regression(Multiple variables of features)

   # y = w1x1 + w1x2 + ........ + wnxn +b

   # y = target variable
   # x = features (independent variable)
   # w = weights(coefficients)
   # b = bias (intercept)

# Performance Metrics

# Mean Squared Error(MSE) - Measures average squared difference between actual and predicted values.(lower is better)
# R-squared Score - Measures how well the model explain variance on the data (Higher is better, max = 1)



# Steps need to follow while applu=ying any machine learning algorithm.


# Step 1: Understand theory  and equation of linear regression.
# Step 2: Import necessary Python libraries
# Step 3: Load the real-world data(Boston Housing Dataset).
# Step 4: EDA (Exploratory Data Analysis) - Pandas analysis and Data Visualization
# Step 5: Preprocess data (Feature Selection, Train-Test-Split).
# Step 6: Train a Linear Regression Model.
# Step 7: Make Predictions.
# Step 8: Evaluate performance (MSE, R-Squared)
# Step 9: Visualize Results
```

### **Practical Session for Linear Regression**

## **Step 2:**

```python
import numpy as np
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split # this is compulsory
from sklearn.linear_model import LinearRegression # algorithm will only be changed here
from sklearn.metrics import mean_squared_error, r2_score # only the matrics name and things will change
```

## **Step 3:**

```python
from sklearn.datasets import fetch_california_housing

california = fetch_california_housing()
df = pd.DataFrame(california.data, columns=california.feature_names)

df['PRICE'] = california.target

df.head()
```

## **Step 5:**

```python
x = df [['MedInc']] # this is my feature
y = df['PRICE'] # this is my target variable i need to predict
```

```python

# Poulation and Smaple

# 100% - Assume (Population)

# X_train, Y_train
# y_train, y_test

# 80% (Train) - 20% (Test)

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

```

## **Step 6:**

```python
 model = LinearRegression()
 model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
# y = mx + c

# Price = m* MedInc + b

print(f"Intercept(b): {model.intercept_}")
print(f"Coefficient(m): {model.coef_[0]}")
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse:.2f}")
print(f"R-squared Score: {r2:.2f}")
```

```python
plt.scatter(X_test, y_test, color='blue', label='Actual Prices')
plt.plot(X_test, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel("Median Income (MedInc)")
plt.ylabel("House Price (in $100,000s)")
plt.title("Linear Regression: MedInc vs PRICE")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('heart_disease_uci.csv')

# Define high-risk patients (age > 55)
df['risk_group'] = df['age'].apply(lambda x: 'High Risk' if x > 55 else 'Low Risk')

# Calculate mean and standard deviation
mean_age = df['age'].mean()
std_age = df['age'].std()

# Plot
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age', kde=True, hue='risk_group', palette={'High Risk':'red', 'Low Risk':'green'}, alpha=0.6)
plt.axvline(mean_age, color='blue', linestyle='--', label='Mean')
plt.axvline(mean_age + std_age, color='purple', linestyle='--', label='Mean + 1 SD')
plt.axvline(mean_age - std_age, color='purple', linestyle='--', label='Mean - 1 SD')
plt.title('Age Distribution with Risk Coloring')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Create the boxplot with jittered data points for smokers vs non-smokers
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Smoker_Status', y='trestbps', palette='pastel')
sns.stripplot(data=df, x='Smoker_Status', y='trestbps', color='black', alpha=0.5, jitter=True)

# Add proper labels and title
plt.title('Resting Blood Pressure: Smokers vs Non-Smokers')
plt.xlabel('Smoker Status')
plt.ylabel('Resting Blood Pressure (trestbps)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
